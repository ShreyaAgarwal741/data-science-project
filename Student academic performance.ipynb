```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, LabelEncoder
from scipy.stats import skew, kurtosis
def load_dataset():
    global df
    df = pd.read_csv("academic_performance_dataset_V2.csv")
    print("Data loaded successfully!\n")
    print(df)
def show_data():
    print("\nFirst 5 rows:\n", df.head())
    print("\nlast five rows:\n",df.tail())
    print("\n summary is:\n",df.info())
def clean_data():
    global df
    print("\nMissing values before cleaning:\n", df.isnull().sum())
    df = df.fillna(df.mean(numeric_only=True))
    print("\nMissing values after cleaning:\n", df.isnull().sum())
def preprocess_data():
    global df
    print("\nPreprocessing...")
    num_cols = df.select_dtypes(include=np.number).columns
    for col in num_cols:
        data = df[col]
        data = data.fillna(data.mean())
        if data.max() == data.min():
            df[col] = 0 
            print(f"{col} skipped (constant column)")
        else:
            df[col] = (data - data.min()) / (data.max() - data.min())
    print("Data preprocessing completed !")
def skew_kurt():
    num_cols = df.select_dtypes(include=np.number).columns
    print("\nSkewness & Kurtosis:\n")
    for col in num_cols:
        data = df[col].dropna()
        if len(data) == 0:
            print(f"{col} -> No data")
            continue
        if data.nunique() == 1:
            print(f"{col} -> Skew: 0.00, Kurtosis: 0.00 (constant data)")
            continue
        sk = skew(data)
        kt = kurtosis(data)
        print(f"{col} -> Skew: {sk:.2f}, Kurtosis: {kt:.2f}")
def probability():
    col = input("Enter column name for probability: ")
    if col in df.columns:
        value = float(input("Enter value: "))
        prob = len(df[df[col] == value]) / len(df)
        print(f"Probability of {value} in {col}: {prob:.4f}")
    else:
        print("Invalid column!")
def visualize():
    print("\nAnalysis data...\n")
    num_cols = df.select_dtypes(include='number').columns
    for col in num_cols:
        plt.figure()
        plt.hist(df[col].dropna(), bins=10)
        plt.title(f"Histogram of {col}")
        plt.xlabel(col)
        plt.ylabel("Frequency")
        plt.show()
while True:
    print("\n===== MENU =====")
    print("1.load dataset")
    print("2. Show Data")
    print("3. Clean Data")
    print("4. Preprocess Data")
    print("5. Skew & Kurtosis")
    print("6. Probability")
    print("7. Visualization")
    print("0. Exit")
    
    choice = input("Enter your choice: ")
    if choice == '1':
        load_dataset()
    if choice == '2':
        show_data()
    elif choice == '3':
        clean_data()
    elif choice == '4':
        preprocess_data()
    elif choice == '5':
        skew_kurt()
    elif choice == '6':
        probability()
    elif choice == '7':
        visualize()
    elif choice == '0':
        print("Exiting...")
        break
    else:
        print("Invalid choice!")
```
